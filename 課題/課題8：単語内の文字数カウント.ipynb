```python
letters = dict.fromkeys (list ("abcdefghijklmnopqrstuvwxyz"), 0)
letters
```

```python
words = []               #words に単語を格納していくつもりです
while True:
  word_new = input(f"英単語を入力してください：")
 
  

  if word_new ==  "":     #""は空欄の意味。空欄のままリターンされたら終わりますでbreak
                          #この行にデバック用として"終了します"を書いておくとよい
    break
  words.append(word_new)   #wordsにword_newに入力された単語を格納せよ
  
  for word in word_new:   #word_newの文字を繰り返しwordに格納し続けなさい。単語の文字を参照しているということかな？
                          #デバック用print(word)
    
    if word in letters:   #wordがlettersに存在するならlettersのwordに数字を足していけ
      letters[word] += 1
                          #デバック用print(letters)

words.sort()              #printの前にwordsをアルファベット順にソートしろ
print(f"入力した英単語：{words}")

for key,value in letters.items():     #lettersのitemsにおいてkeyとvalueを繰り返せ、valueがあるならprintしろ
  if value > 0:
    print(f"{key}が{value}個ありました")
    
```
